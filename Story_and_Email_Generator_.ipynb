```python
from google import genai
from google.genai import types
from google.colab import userdata

client = genai.Client(api_key=userdata.get('GEMINI_API'))
```

```python
MODEL = "gemini-3.1-flash-lite"
```

```python
def generate_story(topic, gener, length_words=300):
    system_prompt = (
        "You are a creative short story writer. "
        "Write an interesting story with a clear beginning, middle, and ending. "
        "Use simple, everyday words that are easy to understand. "
        "Keep sentences short and clear. "
        "Avoid complicated words, jargon, and difficult expressions. "
        "Include interesting characters and a small surprise. "
        "Follow the requested topic, genre, and approximate word count. "
        "Return only the story with a creative title."
    )

    user_prompt = f"""
    Write a short story.
    Topic: {topic}
    Genre: {gener}
    Length: approximately {length_words} words.
    """

    response = client.models.generate_content(
        model=MODEL,
        contents=user_prompt,
        config=types.GenerateContentConfig(
            system_instruction=system_prompt,
            temperature=1,
            max_output_tokens=max(2048, length_words * 3),
            thinking_config=types.ThinkingConfig(thinking_level="low")
        )
    )

    print(response.text)


topic = input("Enter the topic: ")
gener = input("Enter the genre: ")
length_words = int(input("Enter the story length in words: "))

generate_story(topic, gener, length_words)
```

```python
# Email Generator

def generate_email(topic, recipient, tone, length_words=150):
    system_prompt = (
        "You are a helpful email writer. "
        "Write clear emails using simple, everyday words. "
        "Keep sentences short and easy to understand. "
        "Avoid complicated words and jargon. "
        "Include a subject line, greeting, message, and closing. "
        "Follow the requested topic, tone, and approximate word count. "
        "Do not invent details. Use placeholders for missing names or facts. "
        "Return only the email."
    )

    user_prompt = f"""
    Write an email.
    Topic and details: {topic}
    Recipient: {recipient}
    Tone: {tone}
    Length: approximately {length_words} words.
    """

    response = client.models.generate_content(
        model=MODEL,
        contents=user_prompt,
        config=types.GenerateContentConfig(
            system_instruction=system_prompt,
            temperature=1,
            max_output_tokens=max(2048, length_words * 3),
            thinking_config=types.ThinkingConfig(thinking_level="low")
        )
    )

    print(response.text)


topic = input("Enter the email topic and details: ")
recipient = input("Enter the recipient: ")
tone = input("Enter the tone (formal, friendly, polite): ")
length_words = int(input("Enter the email length in words: "))

generate_email(topic, recipient, tone, length_words)
```
